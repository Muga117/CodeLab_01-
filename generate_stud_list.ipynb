```python
#import pandas and logging packages
import pandas as pd
import logging

#configure logging
logging.basicConfig(level=logging.INFO, filename="output_files/studentlog.log", filemode="w") 

#set max columns to 6 and max rows to 200
pd.set_option('display.max_columns', 6)
pd.set_option('display.max_rows', 200)
```

```python
#Get List of Students from File_A and File_B and combine them into one list.
studentsA = pd.read_excel("input_files\\Test Files.xlsx", sheet_name="File_A")
studentsB = pd.read_excel("input_files\\Test Files.xlsx", sheet_name="File_B")
students = pd.concat([studentsA,studentsB])
```

```python
#Count Number of Male and Female Students and output log to studentlog.log
maleCount = students['Gender'].value_counts()["M"]
femaleCount = students['Gender'].value_counts()["F"]
```

```python
#log Number of Male and Female Students
logging.info(f"Number of Male Students: {maleCount}" )
logging.info(f"Number of Female Students: {femaleCount}" )
```

```python
#Define Filters
maleFilter = students['Gender'] == "M"
femaleFilter = students['Gender'] == 'F'
specialFilter = students['Student Name'].str.contains("'")
```

```python
#Remove print() and remain with students[maleFilter] to be able to view output in ipynb
#Output Male Students 
print(students[maleFilter])
```

```python
#Output Female Students 
print(students[femaleFilter])
```

```python
#Output Students whose Names have Special Characters
print(students[specialFilter])
```
